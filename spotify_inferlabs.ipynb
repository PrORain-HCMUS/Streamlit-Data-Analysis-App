```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("BPI\Most Streamed Spotify Songs 2024.csv", encoding="ISO-8859-1")
```

```python
# Combine track + artist for uniqueness
df["Track (Artist)"] = df["Track"] + " - " + df["Artist"].astype(str)

# Convert numbers
cols_to_clean = ["Spotify Streams", "Spotify Playlist Count", "Spotify Playlist Reach",
                 "YouTube Views", "YouTube Likes", "TikTok Posts", "TikTok Likes", "TikTok Views",
                 "YouTube Playlist Reach", "AirPlay Spins", "SiriusXM Spins", "Deezer Playlist Reach",
                 "Pandora Streams", "Pandora Track Stations", "Soundcloud Streams", "Shazam Counts"]

for col in cols_to_clean:
    df[col] = df[col].astype(str).str.replace(",", "").replace("nan", "0")
    df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)

# Parse date
df["Release Date"] = pd.to_datetime(df["Release Date"], errors="coerce")
df = df[df["Release Date"].notnull()]
df.head()
```

```python
# 1. Top 10 songs by Spotify Streams
top_10_streams = df.sort_values(by='Spotify Streams', ascending=False).head(10)
print("Table: Top 10 Songs by Spotify Streams:")
print(top_10_streams[['Track', 'Artist', 'Spotify Streams']])
```

```python
# 2. Top 10 Artists by Spotify Streams
artist_streams = df.groupby('Artist')['Spotify Streams'].sum().sort_values(ascending=False).head(10)
print("\nTable: Top 10 Artists by Spotify Streams:")
print(artist_streams)
```

```python
# 3. Top 10 Albums by Spotify Streams
album_streams = df.groupby('Album Name')['Spotify Streams'].sum().sort_values(ascending=False).head(10)
print("\nTable: Top 10 Albums by Spotify Streams:")
print(album_streams)
```

```python
# 4. Artist with the most songs in the dataset
artist_counts = df['Artist'].value_counts()
most_songs_artist = artist_counts.index[0]
most_songs_count = artist_counts.iloc[0]
print(f"\nNote: The artist with the most songs is {most_songs_artist} with {most_songs_count} songs.")
```

```python
# 5. Oldest song based on the release date
oldest_song = df.sort_values(by='Release Date').iloc[0]
print("\nNote: The oldest song is:")
print(oldest_song[['Track', 'Artist', 'Release Date']])
```

```python
# 6. Average number of streams across all songs
avg_streams = df['Spotify Streams'].mean()
print(f"\nNote: The average number of streams across all songs is: {avg_streams:,.2f}")
```

```python
# 7. Album with the most songs in the dataset
album_counts = df['Album Name'].value_counts()
most_songs_album = album_counts.index[0]
most_songs_album_count = album_counts.iloc[0]
print(f"\nNote: The album with the most songs is {most_songs_album} with {most_songs_album_count} songs.")
```

```python
# 8. Total number of Spotify streams for the first 10 songs
total_streams_first_10 = df['Spotify Streams'].head(10).sum()
print(f"\nNote: The total number of Spotify streams for the first 10 songs is: {total_streams_first_10:,.2f}")
```

```python
# 9. Song with the highest YouTube Views
highest_youtube_views_song = df.sort_values(by='YouTube Views', ascending=False).iloc[0]
print("\nNote: The song with the highest YouTube Views is:")
print(highest_youtube_views_song[['Track', 'Artist', 'YouTube Views']])
```

```python
# 10. Average Spotify Playlist Count per track
avg_playlist_count = df['Spotify Playlist Count'].mean()
print(f"\nNote: The average Spotify Playlist Count per track is: {avg_playlist_count:,.2f}")
```

```python
# 11. Number of songs with more than 1 million Spotify Streams
songs_over_1m = df[df['Spotify Streams'] > 1000000].shape[0]
print(f"\nNote: There are {songs_over_1m} songs with more than 1 million Spotify Streams.")
```

```python
# 12. Median value of Spotify Popularity across all songs
median_popularity = df['Spotify Popularity'].median()
print(f"\nNote: The median value of Spotify Popularity across all songs is: {median_popularity}")
```

```python
# 13. Number of songs released in March
songs_in_march = df[df['Release Date'].dt.month == 3].shape[0]
print(f"\nNote: There are {songs_in_march} songs released in March.")
```

```python
# 14. Artist with the highest average Spotify Playlist Reach per song
artist_playlist_reach = df.groupby('Artist')['Spotify Playlist Reach'].mean().sort_values(ascending=False)
highest_avg_reach_artist = artist_playlist_reach.index[0]
highest_avg_reach = artist_playlist_reach.iloc[0]
print(f"\nNote: The artist with the highest average Spotify Playlist Reach per song is {highest_avg_reach_artist} with {highest_avg_reach:,.2f} reach.")
```

```python
# 15. Number of tracks marked as Explicit
explicit_tracks = df['Explicit Track'].sum()
print(f"\nNote: There are {explicit_tracks} tracks marked as Explicit.")
```

```python
# 16. Top 5 tracks with the highest combined total of TikTok Views and Likes
df['TikTok Total'] = df['TikTok Views'] + df['TikTok Likes']
top_5_tiktok = df.sort_values(by='TikTok Total', ascending=False).head(5)
print("\nTable: Top 5 tracks with the highest combined total of TikTok Views and Likes:")
print(top_5_tiktok[['Track', 'Artist', 'TikTok Views', 'TikTok Likes', 'TikTok Total']])
```

```python
# 17. Track with the highest ratio of Spotify Streams to TikTok Posts
df['StreamsToPostsRatio'] = df['Spotify Streams'] / df['TikTok Posts']
highest_ratio_track = df.sort_values(by='StreamsToPostsRatio', ascending=False).iloc[0]
print("\nNote: The track with the highest ratio of Spotify Streams to TikTok Posts is:")
print(highest_ratio_track[['Track', 'Artist', 'StreamsToPostsRatio']])
```

```python
# 18. Total Soundcloud Streams for the artist with the most tracks
total_soundcloud_streams = df[df['Artist'] == most_songs_artist]['Soundcloud Streams'].sum()
print(f"\nNote: The total Soundcloud Streams for {most_songs_artist} is: {total_soundcloud_streams:,.2f}")
```

```python
# 19. Artist with the widest range in YouTube Views across their tracks
artist_youtube_range = df.groupby('Artist')['YouTube Views'].agg(lambda x: x.max() - x.min()).sort_values(ascending=False)
widest_range_artist = artist_youtube_range.index[0]
widest_range = artist_youtube_range.iloc[0]
print(f"\nNote: The artist with the widest range in YouTube Views is {widest_range_artist} with a range of {widest_range:,.2f}")
```

```python
# 20. Average Spotify Playlist Count for Explicit vs Non-Explicit tracks
avg_playlist_explicit = df.groupby('Explicit Track')['Spotify Playlist Count'].mean()
print("\nTable: Average Spotify Playlist Count for Explicit vs Non-Explicit tracks:")
print(avg_playlist_explicit)
```

```python
# 21. Correlation between Spotify Streams and YouTube Views
correlation = df['Spotify Streams'].corr(df['YouTube Views'])
print(f"\nNote: The correlation between Spotify Streams and YouTube Views is: {correlation:.2f}")
if correlation > 0:
    print("The correlation between Spotify Streams and YouTube Views is positive.")
elif correlation < 0:
    print("The correlation between Spotify Streams and YouTube Views is negative.")
else:
    print("There is no correlation between Spotify Streams and YouTube Views.")

```

```python
# 22. Artist who appears in the most different albums
artist_album_count = df.groupby('Artist')['Album Name'].nunique().sort_values(ascending=False)
most_different_albums_artist = artist_album_count.index[0]
most_different_albums = artist_album_count.iloc[0]
print(f"\nNote: The artist who appears in the most different albums is {most_different_albums_artist} with {most_different_albums} albums.")
```

```python
# 23. Top 3 artists with the highest average TikTok Posts per track
artist_avg_tiktok_posts = df.groupby('Artist')['TikTok Posts'].mean().sort_values(ascending=False).head(3)
print("\nTable: Top 3 artists with the highest average TikTok Posts per track:")
print(artist_avg_tiktok_posts)
```

```python
# 24. Most popular track (based on Spotify Streams) by the artist who appears in the most albums
most_popular_track_by_artist = df[df['Artist'] == most_different_albums_artist].sort_values(by='Spotify Streams', ascending=False).iloc[0]
print(f"\nNote: The most popular track by {most_different_albums_artist} is:")
print(most_popular_track_by_artist[['Track', 'Spotify Streams']])
```

```python
# 25. Total Spotify Streams rank of the artist who released the most tracks in January
df['Release Month'] = df['Release Date'].dt.month
artist_january_releases = df[df['Release Month'] == 1]['Artist'].value_counts()
if not artist_january_releases.empty:
    most_january_artist = artist_january_releases.index[0]
    tracks_by_most_january_artist = df[df['Artist'] == most_january_artist]
    total_streams_rank = tracks_by_most_january_artist['All Time Rank'].sum()
    print(f"\nNote: The total Spotify Streams rank of {most_january_artist} is {total_streams_rank}")
else:
    print("\nNote: No tracks were released in January.")
```

```python
# Hard 1: Track with the highest ratio of Spotify Streams to all platform streams
df['All Platform Streams'] = df['Spotify Streams'] + df['YouTube Views'] + df['Soundcloud Streams'] + df['Pandora Streams']
df['SpotifyToAllRatio'] = df['Spotify Streams'] / df['All Platform Streams']
highest_ratio_track = df.sort_values(by='SpotifyToAllRatio', ascending=False).iloc[0]
print("\nNote: The track with the highest ratio of Spotify Streams to all platform streams is:")
print(highest_ratio_track[['Track', 'Artist', 'SpotifyToAllRatio']])
```

```python
# Hard 2: Artist with the most tracks released on weekends (Saturday or Sunday)
df['DayOfWeek'] = df['Release Date'].dt.dayofweek  # 5=Saturday, 6=Sunday
weekend_releases = df[df['DayOfWeek'].isin([5, 6])]
artist_weekend_counts = weekend_releases['Artist'].value_counts()
most_weekend_artist = artist_weekend_counts.index[0]
most_weekend_count = artist_weekend_counts.iloc[0]
print(f"\nNote: The artist with the most tracks released on weekends is {most_weekend_artist} with {most_weekend_count} tracks.")
```

```python
# Hard 3: Album with the widest distribution of Spotify Playlist Reach (standard deviation)
album_reach_std = df.groupby('Album Name')['Spotify Playlist Reach'].std().sort_values(ascending=False)
widest_reach_album = album_reach_std.index[0]
widest_reach_std = album_reach_std.iloc[0]
print(f"\nNote: The album with the widest distribution of Spotify Playlist Reach is {widest_reach_album} with a standard deviation of {widest_reach_std:,.2f}")
```

```python
# Hard 4: Top 3 oldest explicit songs with above-average Spotify Streams
avg_streams = df['Spotify Streams'].mean()
explicit_above_avg = df[(df['Explicit Track'] == 1) & (df['Spotify Streams'] > avg_streams)]
top_3_oldest_explicit = explicit_above_avg.sort_values(by='Release Date').head(3)
print("\nTable: Top 3 oldest explicit songs with above-average Spotify Streams:")
print(top_3_oldest_explicit[['Track', 'Artist', 'Release Date', 'Spotify Streams']])
```

```python
# Hard 5: Track with the highest TikTok Likes to Spotify Streams ratio (viral potential)
df['TikTokLikesToStreamsRatio'] = df['TikTok Likes'] / df['Spotify Streams']
highest_viral_track = df.sort_values(by='TikTokLikesToStreamsRatio', ascending=False).iloc[0]
print("\nNote: The track with the highest TikTok Likes to Spotify Streams ratio is:")
print(highest_viral_track[['Track', 'Artist', 'TikTokLikesToStreamsRatio']])
```
